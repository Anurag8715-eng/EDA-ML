```python
""""

1. What is Simple Linear Regression?

Simple Linear Regression is a statistical method used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line.
It predicts Y based on the linear equation:
Y = mX + c

2. What are the key assumptions of Simple Linear Regression?

Linearity: Relationship between X and Y is linear.

Independence: Observations are independent.

Homoscedasticity: Constant variance of residuals.

Normality: Residuals follow a normal distribution.

No or minimal outliers.

3. What does the coefficient m represent in the equation Y = mX + c?

The coefficient m is the slope, indicating how much Y changes for a one-unit increase in X.

4. What does the intercept c represent in the equation Y = mX + c?

The intercept c is the value of Y when X = 0.
It represents the baseline level of the dependent variable.

5. How do we calculate the slope m in Simple Linear Regression?

Using the least-squares formula:

𝑚
=
∑
(
𝑋
−
𝑋
ˉ
)
(
𝑌
−
𝑌
ˉ
)
∑
(
𝑋
−
𝑋
ˉ
)
2
m=
∑(X−
X
ˉ
)
2
∑(X−
X
ˉ
)(Y−
Y
ˉ
)
	​

6. What is the purpose of the least squares method in Simple Linear Regression?

The least-squares method minimizes the sum of squared residuals, ensuring the best-fitting straight line.

7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?

R² measures the proportion of variance in Y explained by X.

R² = 1 → Perfect fit

R² = 0 → No linear relationship

8. What is Multiple Linear Regression?

Multiple Linear Regression models the relationship between multiple independent variables and one dependent variable.
General form:
Y = b₀ + b₁X₁ + b₂X₂ + … + bₙXₙ

9. What is the main difference between Simple and Multiple Linear Regression?

Simple: One independent variable

Multiple: Two or more independent variables

10. What are the key assumptions of Multiple Linear Regression?

Linearity between predictors and outcome

Independence of errors

Homoscedasticity

Normality of residuals

No multicollinearity among predictors

11. What is heteroscedasticity, and how does it affect a Multiple Linear Regression model?

Heteroscedasticity occurs when residuals have unequal variance.
Effects:

Coefficients become unreliable

Standard errors become incorrect

Hypothesis tests become invalid

12. How can you improve a Multiple Linear Regression model with high multicollinearity?

Remove highly correlated predictors

Use dimensionality reduction (PCA)

Use regularization (Ridge, Lasso)

Combine correlated variables

13. What are common techniques for transforming categorical variables for use in regression models?

One-Hot Encoding

Label Encoding

Dummy Variables

Target Encoding

14. What is the role of interaction terms in Multiple Linear Regression?

Interaction terms (X₁ × X₂) capture the effect of two variables acting together, improving model accuracy when relationships are not purely additive.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Simple: Y when X = 0

Multiple: Y when all X variables = 0 (often not meaningful)

16. What is the significance of the slope in regression analysis, and how does it affect predictions?

The slope shows how much Y changes for a one-unit change in X.
Higher slope → Stronger impact on predictions.

17. How does the intercept provide context for the relationship between variables?

It defines the starting value of Y and helps position the regression line, giving context to how X values influence Y.

18. What are the limitations of using R² as the sole measure of model performance?

R² always increases with more variables

Does not indicate overfitting

Does not confirm causation

Does not evaluate predictive accuracy

19. How would you interpret a large standard error for a regression coefficient?

A large standard error means the coefficient is unstable, imprecise, and may not be statistically significant.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Identified when residual plot shows:

Funnel shape

Increasing or decreasing spread
Important because it leads to biased standard errors and incorrect conclusions.

21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?

This means the model has too many useless predictors, causing overfitting.
Adjusted R² penalizes irrelevant variables.

22. Why is it important to scale variables in Multiple Linear Regression?

Scaling ensures:

Equal importance of all variables

Improved numerical stability

Better performance for gradient-based algorithms

23. What is polynomial regression?

Polynomial Regression models a nonlinear relationship by adding powers of X (e.g., X², X³) into the equation.

24. How does polynomial regression differ from linear regression?

Linear regression fits a straight line

Polynomial regression fits a curved line, but parameters remain linear

25. When is polynomial regression used?

Used when the data shows curvature and cannot be explained by a straight line.

26. What is the general equation for polynomial regression?
𝑌
=
𝑏
0
+
𝑏
1
𝑋
+
𝑏
2
𝑋
2
+
𝑏
3
𝑋
3
+
…
+
𝑏
𝑛
𝑋
𝑛
Y=b
0
	​

+b
1
	​

X+b
2
	​

X
2
+b
3
	​

X
3
+…+b
n
	​

X
n
27. Can polynomial regression be applied to multiple variables?

Yes.
This is called Multivariate Polynomial Regression, where each variable can have powers and interactions.

28. What are the limitations of polynomial regression?

High risk of overfitting

Very sensitive to outliers

Becomes complex at high degrees

Harder to interpret

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Cross-validation

Adjusted R²

AIC/BIC

Residual analysis

Validation error

30. Why is visualization important in polynomial regression?

Because plots help show:

How well the curve fits the data

Whether overfitting/underfitting occurs

Whether the pattern is truly nonlinear

31. How is polynomial regression implemented in Python?

Typical steps:

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)


This generates polynomial features and fits a regression model.

""""
```
